## Importacion de librerias
En este primer bloque se cargan las herramientas básicas que usaremos en todo el ejercicio.
traemos las librerias: Numpy, Pandas, Matplotmib

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Variables de entrada
En este paso creamos las dos características principales de cada compra: el momento del día y el dinero gastado. Para la hora, elegimos números al azar entre las 0 y las 23 horas, asumiendo que las compras se reparten de manera pareja a lo largo del día.

```python
np.random.seed(42)
n = 300
hora = np.random.randint(0, 24, n)
monto = np.random.exponential(scale=150, size=n) + 10
```

## Logica del fraude
Aquí definimos las reglas para decidir qué tan probable es que una compra sea un fraude. Establecemos que todas las compras tienen un riesgo mínimo del 3%, pero ese riesgo sube drásticamente si la operación ocurre de madrugada o si el monto supera los 400 soles

```python
prob_fraude = 0.03 + 0.35 * ((hora <= 5) | (hora >= 23)) + 0.25 * (monto > 400)
prob_fraude = np.clip(prob_fraude, 0, 0.9)
es_fraude = (np.random.rand(n) < prob_fraude).astype(int)
```

## Estructura de datos en pandas
Una vez que generamos todas las horas, los montos y los resultados de fraude, este bloque los reúne y los organiza en una sola tabla ordenada, similar a una hoja de cálculo.

```python
transacciones = pd.DataFrame({"hora": hora, "monto": monto, "es_fraude": es_fraude})
```

## Estilo visual (colores segun fraude, legitima, fondo) 
En esta parte configuramos el aspecto visual antes de dibujar el gráfico. Asignamos el color verde a las transacciones normales y el rojo a las sospechosas de fraude, además de crear el marco.

```python
colores = {0: "#0ca30c", 1: "#d03b3b"} # verde = legitima, rojo = fraude
etiquetas = {0: "Legitima", 1: "Fraude"}
fig, ax = plt.subplots(figsize=(10, 6), facecolor="#fcfcfb")
ax.set_facecolor("#fcfcfb")
```

## Muestra del grafico
Este bloque es el encargado de pintar los puntos en la pantalla. 
- Primero selecciona únicamente las compras normales y las dibuja como puntos verdes según su hora y su monto.
- luego selecciona los fraudes y los dibuja como puntos rojos sobre el mismo mapa.

```python
for valor in [0, 1]:
    subset = transacciones[transacciones["es_fraude"] == valor]
    ax.scatter(
        subset["hora"], subset["monto"],
        s=70, alpha=0.75,
        c=colores[valor],
        edgecolors="white", linewidths=0.6,
        label=etiquetas[valor],
)
```

## Complementos de la grafica
Consecuentemente se le añade un título explicativo, se le colocan nombres claros a los ejes horizontal y vertical, se agrega una cuadrícula suave para guiarnos con la mirada y se muestra una leyenda en la esquina para recordar qué significa el color verde y el color rojo. Por último, ordena los bordes e imprime la imagen en la pantalla.

```python
ax.set_title("Transacciones por hora y monto", fontsize=15, color="#0b0b0b", pad=14)
ax.set_xlabel("Hora del dia", fontsize=11, color="#52514e")
ax.set_ylabel("Monto (S/)", fontsize=11, color="#52514e")
ax.set_xticks(range(0, 24, 2))
ax.grid(True, color="#e1e0d9", linewidth=0.8)
ax.spines[["top", "right"]].set_visible(False)
ax.legend(title="Tipo de transaccion", frameon=False, loc="upper right")
plt.tight_layout()
plt.show()
```
